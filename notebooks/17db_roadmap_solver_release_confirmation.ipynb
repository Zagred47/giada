## Matrice diagnostica unica
Il run salva gli outcome di rilascio separati dalle metriche del solver e dei candidati. Se si interrompe, scarica comunque l'archivio di log.

```python
OUTPUT=Path('/kaggle/working/artifacts')/f'giada_task17db_solver_release_{REVISION[:7]}_{uuid.uuid4().hex[:8]}'
command=[sys.executable,'-u',str(REPO/'scripts/run_roadmap_task17db.py'),'--repo',str(REPO),'--teacher',str(TEACHER),'--output',str(OUTPUT)]
completed=subprocess.run(command,check=False)
path=OUTPUT/('failure_report.json' if (OUTPUT/'failure_report.json').is_file() else 'final_report.json')
report=json.loads(path.read_text()) if path.is_file() else {'valid':False,'decision':'SUPERVISOR_INTERRUPTED','error':f'exit={completed.returncode}; controllare process.log'}
display({key:report.get(key) for key in ('valid','decision','native_discrete_release_match_ultra','candidate_discrete_release_match_native','reference_stable_at_registered_limits','formula_control_valid','native_pair_count','candidate_pair_count','error_type','error')})
if report.get('valid'):
 rows=[]
 for name in ('native_pairs.json','candidate_pairs.json'):
  for item in json.loads((OUTPUT/name).read_text()):
   rows.append({'source':name,'arm':item.get('arm',item.get('policy')),'seed':item['seed'],'gbar':item['gbar_multiplier'],'max_V_rmse_mV':round(max(v['voltage_rmse_mv'] for v in item['metrics'].values()),5),'release_discrete_match':item['release']['discrete_match']})
 display(rows)
print({'output':str(OUTPUT),'returncode':completed.returncode,'next':'Scarica lo ZIP anche se il processo fallisce'})
```
