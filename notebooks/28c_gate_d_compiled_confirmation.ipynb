# 🧪 GIADA Gate D — conferma compilata simmetrica
Il confronto eager precedente è positivo. Questo controllo separato verifica che il vantaggio permanga quando *entrambi* i rami sono compilati con lo stesso backend. Il motivo della nuova verifica è registrato prima del risultato compilato.


```python
from pathlib import Path
import base64,json,os,shutil,subprocess,sys,uuid
from IPython.display import display,Javascript
WORK=Path('/kaggle/working')/('giada_gate_d_compiled_'+uuid.uuid4().hex[:8]);WORK.mkdir()
REPO=WORK/'giada'
subprocess.run(['git','clone','https://github.com/Zagred47/giada.git',str(REPO)],check=True)
subprocess.run(['git','-C',str(REPO),'fetch','origin',os.environ.get('GIADA_CODE_REVISION','codex/surrogate-validity-audit')],check=True)
subprocess.run(['git','-C',str(REPO),'checkout','--detach','FETCH_HEAD'],check=True)
REVISION=subprocess.check_output(['git','-C',str(REPO),'rev-parse','HEAD'],text=True).strip()
subprocess.run([sys.executable,'-c','import torch; assert torch.cuda.is_available(); print(torch.cuda.get_device_name())'],check=True)
print({'revision':REVISION})
```

```python
OUTPUT=Path('/kaggle/working/artifacts')/('giada_gate_d_compiled_'+REVISION[:7]+'_'+uuid.uuid4().hex[:8])
completed=subprocess.run([sys.executable,'-u',str(REPO/'scripts/run_gate_d_compiled_confirmation.py'),'--output',str(OUTPUT)],check=False)
path=OUTPUT/('failure_report.json' if (OUTPUT/'failure_report.json').exists() else 'final_report.json')
report=json.loads(path.read_text()) if path.is_file() else {'valid':False,'error':f'worker interrupted: {completed.returncode}'}
display({k:report.get(k) for k in ('valid','eager_registered_passed','compiled_confirmation_passed','gate_d_robustly_supported','task29_authorized','error')})
if report.get('rows'):
 display([{k:r[k] for k in ('family','exact_compiled_median_ms','hybrid_compiled_median_ms','paired_compute_reduction_fraction','accuracy_passed')} for r in report['rows']])
print({'output':str(OUTPUT),'exit':completed.returncode})
```
