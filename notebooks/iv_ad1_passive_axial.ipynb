# GIADA IV-A1/A2 → IV-D1 — prerequisito assiale passivo
Questo non è Task 36. Il notebook verifica prima RC/leak/correnti/units nel compartimento singolo e apre solo dopo il gate dei due compartimenti passivi. Nessun modello attivo è giudicato prima di un report IV-D1 valido. Codice, notebook e protocollo sono della stessa revisione.


```python
import json, shutil, subprocess, sys
from pathlib import Path
WORK=Path('/kaggle/working/giada_iv_ad1'); WORK.mkdir(parents=True,exist_ok=True)
REPO=WORK/'giada'
if not REPO.exists(): subprocess.run(['git','clone','--branch','codex/surrogate-validity-audit','https://github.com/Zagred47/giada.git',str(REPO)],check=True)
subprocess.run(['git','-C',str(REPO),'fetch','origin','codex/surrogate-validity-audit'],check=True)
subprocess.run(['git','-C',str(REPO),'checkout','--detach','FETCH_HEAD'],check=True)
try: import neuron
except ImportError: subprocess.run([sys.executable,'-m','pip','install','--quiet','neuron==8.2.7'],check=True); import neuron
assert neuron.__version__.split('+')[0]=='8.2.7',neuron.__version__
REVISION=subprocess.check_output(['git','-C',str(REPO),'rev-parse','HEAD'],text=True).strip()
print({'giada_revision':REVISION,'neuron':neuron.__version__})
```

```python
OUTPUT=Path('/kaggle/working/artifacts/giada_iv_ad1_passive_axial')
assert not OUTPUT.exists(),f'Output già presente: {OUTPUT}; usa una sessione pulita.'
completed=subprocess.run([sys.executable,str(REPO/'scripts/run_iv_ad1_passive_axial.py'),'--output',str(OUTPUT)],cwd=REPO,check=False)
report_file=OUTPUT/'experiment/final_report.json' if (OUTPUT/'experiment/final_report.json').is_file() else OUTPUT/'failure_report.json'
report=json.loads(report_file.read_text()) if report_file.is_file() else {'valid':False,'diagnosis':'NO_REPORT'}
print({'process_exit':completed.returncode,'valid':report.get('valid'),'diagnosis':report.get('diagnosis'),'iv_a1_a2_passed':report.get('iv_a1_a2_passed'),'iv_d1_passed':report.get('iv_d1_passed'),'task36_authorized':report.get('task36_authorized'),'error':report.get('error')})
archive=shutil.make_archive('/kaggle/working/giada_iv_ad1_passive_axial','zip',root_dir=OUTPUT)
print({'archive':archive,'code_revision':REVISION})
```
